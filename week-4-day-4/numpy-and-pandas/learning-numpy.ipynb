### Creating Arrays in Numpy

```python
list1 = [10,20,30]
list1
```

### Using arange to create an array

The difference between array and arange is that array takes in a list and turns it into an array, whereas arange creates an array from a range of numbers

```python
arr2 = np.arange(10, 20, 1) # (start, stop, step-size)
arr2
```

## Multidimensional Arrays

When we combine arrays together, we can add dimensions. Combining two arrays will give us a two-dimensional array. This creates a matrix for us on which we can perform matrix operations.

```python
list2 = [40,50,60]

list3 = [list1, list2]
list3
```

```python
# Create an array from the two joined lists in list3
arr3 = np.array(list3)
arr3
```

```python
# (2, 3) meaning two lists with 3 elements each, this is two dimensional
arr3.shape
```

### Three Dimensional arrays

```python
arr3dim = np.array([[[1,2,3], [3,4,5]], [[5,6,7],[7,8,9]]])
arr3dim
```

```python
arr3dim.shape
```

## Statistical Functions and arrays

```python
# sum
arr3.sum(axis=0)
```

```python
# sum total each row
arr3.sum(1)
```

```python
# mean
arr3.mean()
```

```python
# standard deviation
arr3.std()
```

```python
# Variance
arr3.var()
```

```python
# Min
arr3.min()
```

```python
# Max
arr3.max()
```

## Iterating through arrays

```python
# nd iter = nd iterate
for n in np.nditer(arr3):
    print(n)
```

```python
# If it is an n-dimensional array, we need to iterate through the dimensions first
for dim in arr3:
    for num in dim:
        print(num)
```

## Saving and Loading in Numpy

```python
# to save an array to a file
np.save("new", arr3)
```

```python
# load the file
np.load("new.npy")
```

```python
# Saving as .txt file
np.savetxt("new.txt", arr3, delimiter=",")
```

```python
# Load from .txt file
np.loadtxt("new.txt", arr3, delimiter=",")
```
