### On the iris dataset

```python
# Add a rank column, ranking petal length in asc order
iris["Rank PL"] = iris["petal_length"].rank(ascending=True)
iris
# so for example, the Rank PL is low for row 2 which has a low petal length of 1.3, the decimals
# of the Rank PL indicates the ties and multiple of the same type of petal length so their rank is grouped
```

```python
# Find the mean of each column for each species
iris.groupby("species").mean()
# so originally I thought I had already done this through iris.describe(), but no that calculates the
# mean of the whole dataset, whereas this question asks for each species, so I need to group by first
# to split the data into species, then calculate the means
```

```python
# Create a new df for each species separately (3 total)
setosa_df = iris[iris['species'] == 'setosa']
versicolor_df = iris[iris['species'] == 'versicolor']
virginica_df = iris[iris['species'] == 'virginica']
# it seems hard to remember but it's easier than you think: the inside brackets are just checking if the species
# is the one you need from the iris dataset, then the outside ones are just asking to return that one and assign
# it a name for the new dataframe
setosa_df.head(10) # it has 49 rows so I'm using head to only show 10
```

```python
# Create 2 final df, one for sepal info and one for petal info (with data from all three species in each)
sepal_df = iris[['sepal_length', 'sepal_width', 'species']]
petal_df = iris[['petal_length', 'petal_width', 'species']]
# okay so the reason it is not in single brackets is because otherwise it would be looking for one column named
# those three parameters, and won't find it, the second set of square brackets shows that it is a list of columns
petal_df.head(5)
```

### Merging

```python
# Find a way to merge your sepal_info and petal_info dataframes
merged_df = pd.concat([sepal_df, petal_df], axis=1) # the reason for axis=1 is so that pandas merges them
# column wise, rather than it defaulting to axis=0 and now petal_df would be underneath sepal_df vertically
merged_df.head()
```

```python
# Drop repeated columns from your merged df
# but as you can see above this creates two columns of the same stuff
merged_df = merged_df.loc[:, ~merged_df.columns.duplicated()]
# okay this one is hella confusing, basically you are using loc to select the columns, the : means keep all rows,
# the ~ means NOT, so basically without it you would mark all unique columns as false which would delete them and
# only leave the duplicate, so you need it, and the rest is just using the duplicated method
# also remember you aren't deleting the old duplicate column, only selecting the ones you want to keep
merged_df.head()
```
