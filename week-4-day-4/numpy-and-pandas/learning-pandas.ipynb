# Learning Pandas

Pandas is a high-level data manipulation package which was built on top of Numpy. The key structures within pandas include Series and DataFrames

```python
import numpy as np
import pandas as pd
```

## Series

A Series is a 1d ndarray with axis labels (an index)

```python
# Creating a series from a list
x = pd.Series([10,20,30,40,50])
x
```

```python
# Accessing the index
x.index
```

```python
# Accessing the values
x.values
```

```python
# Accessing the data type of the Series
x.dtype
```

```python
# Creating a Series with a custom index
data = [450, 650, 870]
Sales = pd.Series(data, index=["Don", "Mike", "Edwin"])
Sales
```

```python
# Check the type
type(Sales)
```

```python
Sales.index
```

```python
# Using index names
Sales["Don"]
```

```python
# We can still use positional index as well
Sales[0]
```

## DataFrames

DataFrames are two-dimensional (2d), size-mutable, potentially heterogeneous (diverse) tabular data structures. This data structure contains labeled axes (rows and columns).

```python
### Creating a DataFrame (DF) from a list
df_data = [["Adrian", 20], ["Bethany", 23], ["Chloe", 41]]

# When creating the df, we need to give it the 2d array and the columns names we want it to use
df = pd.DataFrame(df_data, columns=["Name", "Age"])
df
```

```python
# Creating a DataFrame from a dictionary
new_dict = {'Name':['Tom','Jane','Steve','Lucy'],'Sales':[250,300,350,400]}

df_dict = pd.DataFrame(new_dict)
df_dict
```

```python
# Adding a custom index
df_dict_index = pd.DataFrame(new_dict, index=["rank1", "rank2", "rank3", "rank4"])
df_dict_index
```

```python
# Creating a DataFrame from a list of Dictionaries
# This is the same data as we had previously, but in a different format
# This is useful when working with json
dict_list = [{'Name':'Tom','Sales':250},{'Name':'Jane','Sales':300},{'Name':'Steve','Sales':350},
             {'Name':'Lucy','Sales':400}]

df_dict_list = pd.DataFrame(dict_list)
df_dict_list
```

```python
# Creating a df from a dictionary of Series
east = pd.Series([1000, 1200, 3400], index=["Q1", "Q2", "Q3"])
east
```

```python
west = pd.Series([1100, 1300, 2400, 3500], index=["Q1", "Q2", "Q3", "Q4"])
west
```

```python
# We can combine Series into a singlular df
df_region = pd.DataFrame({"East":east, "West":west})
df_region
```

```python
# We can add more Series to the df, after it is created
df_region["North"] = [2000,3000,2500,4000]
df_region["South"] = [1500,2000,1500,4000]
df_region
```

```python
years = ["2016", "2017", "2018", "2019"]
df_region["Years"] = years
df_region
```

```python
# Make years the index rather than a column
df_region = df_region.set_index("Years")
df_region
```

```python
new_df = df_region.reindex(["2017", "2018", "2019", "2020", "2021"])
new_df
```

```python
## reindex columns
re_indexed = new_df.reindex(columns=["North", "East", "South", "New"])
re_indexed
```

### Missing Data

```python
# Filling in missing values
# We may want to change all NaN values to 0 (or some specific number)
# This will be especially useful when working with certain types of algorithms
# Some algorithms cannot deal with NaN values
 
re_indexed.fillna(0) # note we are not actually assigning this to a new dataframe, this is temporary
```

```python
# If we don't want to fill with a single value, we can change the fill method
# methods we can pick from - ‘backfill’, ‘bfill’, ‘pad’, ‘ffill’
# These methods will take the last available value and carry it to the next item
re_indexed.ffill()
```

```python
# Default method for interpolation is linear, this can be changed.
re_indexed.interpolate()
```

### Dropping items in Dataframes

```python
# Sometimes we would rather drop than fill
re_indexed.dropna()
```

```python
# We can specify columns and methods when dropping with .dropna()
# axis 1 = columns, 0 = rows
# how = all, needs all values to NaN, other option is "any"
re_indexed.dropna(axis=1, how="all")
```

```python
# You can also set a threshold for dropping rows/columns
re_indexed.dropna(thresh=2)
```

```python
re_indexed.dropna(thresh=3)
```

```python
# Dropping based on index
re_indexed.drop("2019")
```

```python
# We can easily check for and remove duplicated rows
# We need to create a DataFrame with duplicates first to demonstrate this
df_dup = pd.DataFrame([['A',1],['B',2],['A',1]])
df_dup
```

```python
# Find duplicate rows
df_dup.duplicated()
```

```python
# drop duplicate rows
# inplace = True, will make changes to the df permanent
df_dup.drop_duplicates(inplace=True)
```

```python
df_dup.duplicated()
```

### Data alignment

We can add Series and columns in DataFrames together. Since Series are made up of singular columns, it is easy to add them togehter, with DataFrames there is an additional step of specifying the column to add. These examples will add DataFrame columns (Note that for Series, this will work the same, without having to input the column names).

```python
# Adding two columns in a df
new_df["East"] + new_df["North"]
```

```python
# we can use .add to specify the value to be used in case of a NaN value
new_df["East"].add(new_df["North"], fill_value=0)
```

### Index Heirarchy

```python
# Creating a DataFrame with a hierarchical index
# We are using numpy arange to create a range of numbers, and re-shaping it so that the numbers will fit on a 4x4 grid
df_temp = pd.DataFrame(np.arange(16).reshape(4,4), index = [['2018','2018','2019','2019'],['Jan','Feb','Jan','Feb']],
                   columns = [['NY','NY','LA','LA'],['Cold','Hot','Cold','Hot']])
df_temp
```

```python
# We can add index and column names
df_temp.index.names=["Year", "Month"]
df_temp.columns.names=["State","Temp"]
df_temp
```

```python
# Swap levels of a particular axis
#axis=0 --> index axis=1 --> columns
df_temp.swaplevel("State","Temp", axis=1)
```

```python
# Sorting on the first level of the index
df_temp.sort_index(level=0)
```

```python
# Sorting on the second level of the index
df_temp.sort_index(level=1)
```

```python
# Accessing singular columns in a multi-index df
df_temp["NY"]["Cold"]
```

```python
# You can also use iloc
df_temp.iloc[0]
```

```python
# using loc for the same thing, Jan 2018
df_temp.loc["2018", "Jan"]
```
