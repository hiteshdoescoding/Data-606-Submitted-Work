## Weather Data

```python
temperatures = np.array([
    18, 21, 25, 17, 16,
    23, 28, 30, 22, 19,
    15, 24, 27, 20
])
```

Second task:

Days above 25 degrees
Days below 20 degrees
Avg temp
Hottest day
Coldest day

How many days exceeded the avg temp?

```python
# Days above 25 degrees
np.sum(temperatures > 25) # this answer means 3 days, the int64 just means that the value was stored as a 64-bit integer
```

```python
# Days below 20 degrees
np.sum(temperatures < 20)
```

```python
# Average temperature
np.mean(temperatures)
```

```python
# Hottest day
np.max(temperatures)
```

```python
# Coldest day
np.min(temperatures)
```

```python
# How many days exceeded the average temperature?
np.sum(temperatures > np.mean(temperatures))
# find the mean which we had previously
# next compare every number in the array to the average temperature, as true and false
# then the sum is the number of true values
```
